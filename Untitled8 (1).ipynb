Now, I will train a `RandomForestRegressor` for temperature, another `RandomForestRegressor` for wind speed, and a `RandomForestClassifier` for the weather condition. After training, I'll evaluate their performance using Mean Absolute Error (MAE) for regression and Accuracy for classification.

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error, accuracy_score

# --- Temperature Model (Regression) ---
temp_model = RandomForestRegressor(random_state=42)
temp_model.fit(X_train, y_temp_train)
y_temp_pred = temp_model.predict(X_test)
mae_temp = mean_absolute_error(y_temp_test, y_temp_pred)
print(f"Temperature Model MAE: {mae_temp:.4f}")

# --- Wind Speed Model (Regression) ---
wind_model = RandomForestRegressor(random_state=42)
wind_model.fit(X_train, y_wind_train)
y_wind_pred = wind_model.predict(X_test)
mae_wind = mean_absolute_error(y_wind_test, y_wind_pred)
print(f"Wind Speed Model MAE: {mae_wind:.4f}")

# --- Condition Model (Classification) ---
condition_model = RandomForestClassifier(random_state=42)
condition_model.fit(X_train, y_condition_train)
y_condition_pred = condition_model.predict(X_test)
accuracy_condition = accuracy_score(y_condition_test, y_condition_pred)
print(f"Condition Model Accuracy: {accuracy_condition:.4f}")
```

### Step 1: Redefine Features and Target for District Prediction

Now, I'll set up the data for predicting `district`. The input features (`X`) will be `year`, `month`, `temperature`, `condition_encoded`, `rainfall_mm`, and `humidity`. The target (`y`) will be `district_encoded`.

```python
from sklearn.model_selection import train_test_split

# Define new features (X) and target (y) for district prediction
X_district = df_encoded[['year', 'month', 'temperature', 'condition_encoded', 'rainfall_mm', 'humidity']]
y_district = df_encoded['district_encoded']

# Split the data into training and testing sets (80% train, 20% test)
X_district_train, X_district_test, y_district_train, y_district_test = train_test_split(
    X_district, y_district, test_size=0.2, random_state=42, stratify=y_district
)

print(f"Training set size for district model: {len(X_district_train)} samples")
print(f"Testing set size for district model: {len(X_district_test)} samples")
```

### Step 2: Train and Evaluate District Prediction Model

I will now train a `RandomForestClassifier` to predict the district and evaluate its performance using accuracy.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Initialize the RandomForestClassifier model for district prediction
district_model = RandomForestClassifier(random_state=42)

# Train the model using the training data
district_model.fit(X_district_train, y_district_train)

# Make predictions on the test set
y_district_pred = district_model.predict(X_district_test)

# Calculate the accuracy of the model
accuracy_district = accuracy_score(y_district_test, y_district_pred)
print(f"District Model Accuracy: {accuracy_district:.4f}")
```
